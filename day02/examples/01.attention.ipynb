## 6. AI도 모든 곳을 똑같이 참고하지 않았습니다

- AI의 답은 **수신기** 입니다. 우리가 찾은 답과 같습니다.
- 그림에서 막대 하나만 뚜렷하게 높고, 나머지는 거의 보이지 않을 만큼 낮습니다.
- 가장 높은 곳은 `수신기가` — 우리가 답을 찾을 때 사용했던 바로 그 부분입니다.
- 주문 · 택배 · 기사님 · 영수증 부분은 상대적으로 훨씬 낮습니다.

즉 이 교육용 모델은 메시지의 **모든 부분을 똑같이 참고하지 않았습니다.**

**이 그림을 읽는 방법**

> 이 그림은 이번 계산에서 어느 부분의 정보가 **상대적으로 더 많이 반영되었는지**를
> 쉽게 보여 주는 그림입니다.
> AI가 그 단어의 뜻을 사람처럼 이해했다는 뜻은 아닙니다.

<br>

그런데 여기서 의심이 하나 듭니다.

**혹시 AI가 마지막 문장을 통째로 외운 것은 아닐까요?**
아니면 `수신기` 라는 말만 찾도록 배운 것은 아닐까요?

## 7. 문맥을 바꿔 보겠습니다

확인해 보기 위해 메시지 두 개를 준비했습니다.

```
[문맥 A]  수신기는 정상입니다  충전기는 정상입니다  설명서가 빠져 있었습니다  이것만 다시 보내 주세요

[문맥 B]  수신기가 빠져 있었습니다  충전기는 정상입니다  설명서는 정상입니다  이것만 다시 보내 주세요
```

**같게 맞춘 것**

- 나오는 물건 — 수신기 · 충전기 · 설명서 (나오는 순서까지 같습니다)
- 표현 방식 — `…는 정상입니다` / `…가 빠져 있었습니다`
- 마지막 요청 — `이것만 다시 보내 주세요` (글자 하나까지 같습니다)
- 단어 수 — 두 메시지 모두 11개

**달라진 것**

- 어느 물건이 빠졌는가 — 문맥 A는 **설명서**, 문맥 B는 **수신기**
- 그에 따라 `수신기는` ↔ `수신기가` 처럼 뒤에 붙는 조사도 함께 바뀝니다.

<br>

## 8. 같은 요청인데 참고한 곳이 달라졌습니다

| | 문맥 A | 문맥 B |
|---|---|---|
| 빠진 물건 | 설명서 | 수신기 |
| 더 많이 참고한 곳 | `설명서가` | `수신기가` |
| AI의 답 | 설명서 | 수신기 |

마지막 요청은 글자 하나까지 같았는데, 더 많이 참고한 곳은 반대편으로 옮겨 갔습니다.

- `이것` 이 무엇을 가리키는지는 **앞의 내용에 따라 달라집니다.**
- 그래서 문맥이 달라지면 **이번 계산에서 더 많이 참고하는 곳도 달라질 수 있습니다.**
- 참고할 곳은 미리 정해져 있는 것이 아니라, 메시지마다 그때그때 다시 정해집니다.

## 9. 오늘의 정리

```
지금 해결할 문제가 있다
        ↓
문장에는 관련 있는 부분과 관계없는 부분이 함께 있다
        ↓
AI는 관련 있는 부분을 상대적으로 더 많이 참고할 수 있다
        ↓
참고한 내용을 바탕으로 다음 처리를 한다
        ↓
문맥이 달라지면 더 많이 참고하는 부분도 달라질 수 있다
```

이것이 **Attention** 의 핵심 아이디어입니다.

### 업무에서는 어디에 쓰일까요?

생성형 AI는 관련 있는 문맥을 참고하며 처리합니다.
