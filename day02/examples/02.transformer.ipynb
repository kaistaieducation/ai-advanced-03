## 9. 오늘 실습에서 Transformer 가 한 일

```
같은 단어가 들어 있다고 해서 같은 질문은 아니다
        ↓
문장 안의 위치와 관계를 함께 반영한다
        ↓
문장 전체의 패턴을 바탕으로
        ↓
관련 가능성이 높은 안내문 후보를 추천한다
```

그리고 오늘 함께 본 또 하나의 사실이 있습니다.

> **교육용으로 만든 모델은 처음 보는 표현에서 틀리기도 합니다.**

그래서 모델이 하는 일은 **후보를 먼저 보여 주는 것** 까지이고,
최종 판단과 회신은 **사람이 확인** 합니다.

### 업무에서는 어디에 쓰일까요?

들어온 문서나 문의를 **성격별로 나누어 담당자에게 보내는 일**,
비슷해 보이는 요청을 **미리 갈라 놓는 일** 에 이런 방식이 쓰입니다.
